```python
import urllib, requests, socket, re, lxml, io, bs4, sqlite3, pandas, sqlalchemy
from bs4 import BeautifulSoup
from urllib import request
from urllib.parse import urljoin, unquote
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from geopy.distance import distance
from geopy.geocoders import Nominatim
import csv
url = "http://www.lovrikinfo.ru/metrogps.php?ysclid=mp9rm5jncn754073184"
# 1. Получаем штмл
response = requests.get(url)
html_text = response.content

# 2. Передаём текст в BeautifulSoup
soup = BeautifulSoup(html_text, 'html.parser')

# 3.  код для парсинга таблицы
stations = []
for tr in soup.find_all('tr'):
    cells = tr.find_all('td')
    if len(cells) == 3:
        name = cells[0].get_text(strip=True)
        lat = cells[1].get_text(strip=True)
        lon = cells[2].get_text(strip=True)
        try:
            float(lat)
            float(lon)
            stations.append((name, lat, lon))
        except ValueError:
            continue

print(f"Спарсено {len(stations)} станций")

# Сохраняем результат
with open('moscow_metro.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow(['station', 'latitude', 'longitude'])
    writer.writerows(stations)

print("Данные сохранены в moscow_metro.csv")

#------
# 1. Загружаем список станций метро (два столбца: lat, lon)
metro = pd.read_csv('moscow_metro.csv')

# 2. Получаем координаты дома через нашу библиотеку
geolocator = Nominatim(user_agent="house_analyzer")
home = geolocator.geocode("Москва, ул. Профсоюзная,20")
home_coords = (home.latitude, home.longitude)

# 3. Ищем ближайшую станцию
min_dist = float('inf')
closest_station = None
for _, row in metro.iterrows():
    station_coords = (row['latitude'], row['longitude'])
    dist = distance(home_coords, station_coords).meters
    if dist < min_dist:
        min_dist = dist
        closest_station = row['station']

print(f"Ближайшее метро: {closest_station}, {min_dist:.0f} метров")
print(home_coords)
```

```python
import urllib, requests, socket, re, lxml, io, bs4, sqlite3, pandas, sqlalchemy
from bs4 import BeautifulSoup
from urllib import request
from urllib.parse import urljoin, unquote
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from geopy.distance import distance
from geopy.geocoders import Nominatim
import csv

geolocator = Nominatim(user_agent="house_analyzer")
home = geolocator.geocode("Москва, рязанский проспект, 1")
home_coords = (home.latitude, home.longitude)
lat, lon = home.latitude, home.longitude
#  задаем объекты в виде точки, линии и области (улицы скверы бульвары и т д) и указываем радиус объктов вокруг (напр 1500м)
query = f"""
[out:json];
(
  node["leisure"="park"](around:1500,{lat},{lon});
  way["leisure"="park"](around:1500,{lat},{lon});
  relation["leisure"="park"](around:1500,{lat},{lon});
);
out center;
"""

headers = {
    "Accept": "application/json",
    "User-Agent": "MyHouseAnalyzer/1.0 (test)"
}

# ищем данные с геосайта и применяем заготовленный запрос выше
response = requests.get(
    "https://overpass.openstreetmap.fr/api/interpreter",
    params={'data': query},
    headers=headers
)

print(f"Статус: {response.status_code}")
if response.status_code == 200:
    elements = response.json().get('elements', [])
    print(f" Найдено парков: {len(elements)}")
    print("\n".join([f"   - {el.get('tags', {}).get('name', 'без названия')}" for el in elements]))
    for el in elements[:5]:
        name = el.get('tags', {}).get('name', 'без названия')
        print(f"   - {name}")
else:
    print(f" Ошибка: {response.status_code}")
```
